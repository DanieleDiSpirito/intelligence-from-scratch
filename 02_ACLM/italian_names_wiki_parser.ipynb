```python
import requests
```

## Masculine names

```python
m_names = []

session = requests.Session()

session.headers.update({
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
})

url = "https://commons.wikimedia.org/w/api.php"

params = {
    "action": "query",
    "list": "categorymembers",
    "cmtitle": "Category:Italian-language_masculine_given_names",
    "cmlimit": "500",
    "format": "json"    
}

page_count = 1

while True:
    print(f"Fetching page {page_count}... ({len(m_names)} names collected so far)")
    response = session.get(url, params=params).json()
    m_names.extend([x["title"].split(":")[1].split(" ")[0] for x in response['query']['categorymembers'][1:]])
    if 'continue' not in response:
        break
    params['cmcontinue'] = response['continue']['cmcontinue']
    page_count += 1
    
print(f"Total number of names: {len(m_names)}")
```

```python
with open("italian_masculine_given_names.txt", "w") as f:
    for name in m_names:
        f.write(name + "\n")
```

### Common names $\rightarrow$ citations $\geq 15$

```python
session = requests.Session()
session.headers.update({
    "User-Agent": "NameCitationScraper/1.0"
})

url = "https://commons.wikimedia.org/w/api.php"

params = {
    "action": "query",
    "generator": "categorymembers",
    "gcmtitle": "Category:Italian-language_masculine_given_names",
    "gcmlimit": "500",
    "prop": "categoryinfo",         # This gives us the count of files/subcats
    "format": "json"
}

results = {}

while True:
    response = session.get(url, params=params).json()
    pages = response.get('query', {}).get('pages', {})
    for page_id, info in pages.items():
        title = info["title"]
        if ":" in title:
            name = title.split(":")[1].split(" ")[0]
            citations = info.get("categoryinfo", {}).get("subcats", 0)
            if citations >= 15:
                results[name] = citations
    print(f"Processed {len(results)} names...")
    if 'continue' in response:
        # Important: generator continuation uses 'gcmcontinue'
        params['gcmcontinue'] = response['continue']['gcmcontinue']
    else:
        break

sorted_results = dict(sorted(results.items(), key=lambda item: item[1], reverse=True))

print("\nTop 10 Names by Citation Count:")
for name, count in list(sorted_results.items())[:10]:
    print(f"{name:15}: {count:4} files")
```

```python
with open("italian_popular_masculine_names.txt", "w") as f:
    for name, count in sorted_results.items():
        f.write(f"{name}\n")
```
